```python
import matplotlib.pyplot as plt
```

```python
def direct_line(x1,y1,x2,y2):
  pixels = []
  print("\n --Direct line equation method--")
  print("Starting Point: ",(x1,y1))
  print("Ending Point: ",(x2,y2))
  dx = x2-x1
  dy = y2-y1 # Corrected: should be y2-y1

  if dx==0: # Vertical Line
    print("case: Vertical Line.\n")
    print("slope is undefined.\n")
    start_y = min(y1,y2)
    end_y = max(y1,y2)
    for y in range(start_y, end_y+1):
      pixels.append((x1,y))
      print(
          "x=",x1,
          "y=",y,
          "pixels ->",(x1,y)
      )
  else:
    m = dy/dx
    b = y1 - m*x1
    print("Slope: ",m)
    print("Y-intercept: ",b)

    if abs(m)<=1: # Shallow Line
      print("case: Shallow Line.\n")
      start_x = min(x1,x2)
      end_x = max(x1,x2)
      for x in range(start_x, end_x+1):
        y = int(m*x+b)
        pixels.append((x,y))
        print(
            "x=",x,
            "y=",y, # Corrected: round_y was not defined, use y directly
            "pixels ->",(x,y)
        )
    else: # Steep Line
      print("Case: Steep Line.\n")
      start_y = min(y1,y2)
      end_y = max(y1,y2)
      for y in range(start_y, end_y+1):
        x = (y-b)/m
        round_x = round(x)
        pixels.append((round_x,y))
        print(
            "x=",round_x,
            "y=",y,
            "pixels ->",(round_x,y)
        )
  return pixels
```

```python
x1 = int(input("Enter the value of x1: "))
y1 = int(input("Enter the value of y1: "))
x2 = int(input("Enter the value of x2: "))
y2 = int(input("Enter the value of y2: "))

pixels = direct_line(x1,y1,x2,y2)

print(pixels)

```

```python
## Plotting Pixels ##

# Extract x and y coordinates from the pixels list
x_pixels = [p[0] for p in pixels]
y_pixels = [p[1] for p in pixels] # Correctly define y_pixels

plt.figure(figsize=(8,6))

# Plot the ideal mathematical line
plt.plot(
    [x1,x2],
    [y1,y2],
    label = "Ideal Mathematical Line"
)

# Scatter plot the generated pixels
plt.scatter(
    x_pixels,
    y_pixels,
    s = 100, # Use 's' for size in scatter
    marker="s",
    color="red",
    label = "Generated Pixels"
)

# Add text labels for each pixel
for x,y in pixels:
  plt.text(
      x + 0.08, # Offset x for better visibility
      y + 0.08, # Offset y for better visibility
      f"({x},{y})",
      fontsize = 10
  )

# Set plot aesthetics (should be outside any pixel loop)
plt.xticks(range(min(x1,x2)-1,max(x1,x2)+2))
plt.yticks(range(min(y1,y2)-1,max(y1,y2)+2))
plt.grid(True)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Direct Line Algorithm")
plt.legend()
plt.show() # Call show once at the end
```
