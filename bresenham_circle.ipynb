```python
import matplotlib.pyplot as plt
```

```python
def bresenham_circle(xc, yc, r):

    pixels = []

    print("\n--- Bresenham's Circle Drawing Algorithm ---")
    print("Center:", (xc, yc))
    print("Radius:", r)

    x = 0
    y = r

    # Initial decision parameter
    d = 3 - 2 * r

    while x <= y:

        # 8-way symmetry
        points = [
            (xc + x, yc + y),
            (xc - x, yc + y),
            (xc + x, yc - y),
            (xc - x, yc - y),

            (xc + y, yc + x),
            (xc - y, yc + x),
            (xc + y, yc - x),
            (xc - y, yc - x)
        ]

        # Store pixels
        for point in points:
            if point not in pixels:
                pixels.append(point)

        print(
            "x =", x,
            "y =", y,
            "Decision Parameter =", d
        )

        # Update decision parameter
        if d < 0:
            d = d + 4 * x + 6

        else:
            d = d + 4 * (x - y) + 10
            y = y - 1

        x = x + 1

    return pixels
```

```python
xc = int(input("Enter center x: "))
yc = int(input("Enter center y: "))
r = int(input("Enter radius: "))
```

```python
pixels = bresenham_circle(xc, yc, r)

print("\nGenerated Pixels:")
print(pixels)
```

```python
x_pixels = [p[0] for p in pixels]
y_pixels = [p[1] for p in pixels]

plt.figure(figsize=(8, 8))

# Generated circle pixels
plt.scatter(
    x_pixels,
    y_pixels,
    s=100,
    marker="s",
    color="red",
    label="Bresenham Circle Pixels"
)

# Center point
plt.scatter(
    [xc],
    [yc],
    s=150,
    marker="o",
    color="green",
    label="Center"
)

# Ideal mathematical circle
circle = plt.Circle(
    (xc, yc),
    r,
    fill=False,
    label="Ideal Circle"
)

plt.gca().add_patch(circle)

# Label pixels
for x, y in pixels:
    plt.text(
        x + 0.1,
        y + 0.1,
        f"({x},{y})",
        fontsize=8
    )

# Axis settings
plt.xticks(
    range(xc - r - 1, xc + r + 2)
)

plt.yticks(
    range(yc - r - 1, yc + r + 2)
)

plt.grid(True)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.title("Bresenham's Circle Drawing Algorithm")

plt.axis("equal")

plt.legend()

plt.show()
```
